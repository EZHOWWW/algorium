# Getting Started with the Optimization Library

```python
import sys
sys.path.append('../')
from opt_library.core.centralized.zeroth_order.random_search.random_search import RandomSearch
from opt_library.simulator.problems import Sphere
```

```python
problem = Sphere(dim=10)
optimizer = RandomSearch(num_samples=1000)
solution, fitness = optimizer.fit(problem, budget=1000, seed=42)
print(f'Best solution found: {solution}
')
print(f'Best fitness: {fitness}')
```
