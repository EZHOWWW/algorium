# Getting Started with the Optimization Library

```python
import sys
import os
import numpy as np

# Get the absolute path to the project root
project_root = os.path.abspath(os.path.join(os.getcwd(), "../.."))
sys.path.append(project_root)
from opt_library.core.centralized.zeroth_order.random_search.random_search import RandomSearch
from opt_library.core.centralized.first_order.stochastic_gradient_descent.stochastic_gradient_descent import StochasticGradientDescent
from opt_library.core.centralized.zeroth_order.zeroth_order_sgd.zeroth_order_sgd import ZOSGD
from opt_library.simulator.problems import Sphere, LinearRegressionProblem
```

```python
problem = Sphere(dim=10)
optimizer = RandomSearch(num_samples=1000)
solution, fitness = optimizer.fit(problem, budget=1000, seed=42)
print(f'Best solution found: {solution}')
print(f'Best fitness: {fitness}')
```

## Stochastic Gradient Descent (SGD)

```python
np.random.seed(42)
n_samples, n_features = 100, 5
X = np.random.rand(n_samples, n_features)
true_w = np.random.randn(n_features)
y = X @ true_w + 0.1 * np.random.randn(n_samples)

problem = LinearRegressionProblem(X, y)
optimizer = StochasticGradientDescent(learning_rate=0.1, batch_size=16, epochs=100)
starting_weights = np.zeros(n_features)

final_w, final_value = optimizer.fit(problem=problem, starting_point=starting_weights)
print(f'Final weights: {final_w}')
print(f'Final value: {final_value}')
```

## Zeroth-Order Stochastic Gradient Descent (ZO-SGD)

```python
problem = Sphere(dim=10)
optimizer = ZOSGD(learning_rate=0.1, mu=0.01, batch_size=4)
starting_point = np.ones(10) * 5.0

class MaxIterStoppingCondition:
    def __init__(self, max_iter=100):
        self.max_iter = max_iter
        self.iter = 0

    def should_stop(self, *args, **kwargs):
        self.iter += 1
        return self.iter > self.max_iter

stopping_condition = MaxIterStoppingCondition(max_iter=100)

final_x, final_value = optimizer.fit(
    problem=problem,
    starting_point=starting_point,
    stopping_condition=stopping_condition,
)

print(f'Final solution: {final_x}')
print(f'Final value: {final_value}')
```
